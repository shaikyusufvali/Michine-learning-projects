```python
import pandas as pd
import numpy as np
```

```python
train_df = pd.read_csv("../data/raw/train_data.csv")

train_df.head()
```

```python
print("Rows:", train_df.shape[0])
print("Columns:", train_df.shape[1])
```

```python
train_df.isnull().sum()
```

```python
print("Duplicate rows:", train_df.duplicated().sum())
```

```python
train_df = train_df.drop_duplicates()

print("Shape after removing duplicates:", train_df.shape)
```

```python
train_df.dtypes
```

```python
train_df["departure_hour"] = pd.to_datetime(
    train_df["scheduled_departure"],
    format="%H:%M"
).dt.hour
```

```python
train_df[["scheduled_departure", "departure_hour"]].head()
```

```python
train_df = train_df.drop(
    columns=["scheduled_departure"]
)
```

```python
train_df[[
    "distance_km",
    "temperature",
    "rainfall",
    "station_delay",
    "previous_delay",
    "delay_minutes"
]].describe()
```

```python
numeric_columns = [
    "distance_km",
    "temperature",
    "rainfall",
    "station_delay",
    "previous_delay",
    "delay_minutes"
]

for column in numeric_columns:
    print(
        column,
        "negative values:",
        (train_df[column] < 0).sum()
    )
```

```python
print(train_df.info())
print("\nMissing values:")
print(train_df.isnull().sum())
```

```python
train_df.to_csv(
    "../data/processed/delay_processed.csv",
    index=False
)

print("Train delay dataset cleaned and saved successfully!")
```
