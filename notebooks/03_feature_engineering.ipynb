```python
import pandas as pd
import numpy as np
```

```python
delay_df = pd.read_csv(
    "../data/processed/delay_processed.csv"
)

delay_df.head()
```

```python
weather_mapping = {
    "Clear": 0,
    "Cloudy": 1,
    "Rain": 2,
    "Heavy Rain": 3,
    "Fog": 4
}

delay_df["weather_encoded"] = delay_df["weather"].map(
    weather_mapping
)
```

```python
delay_df[
    ["weather", "weather_encoded"]
].head()
```

```python
delay_df["route"] = (
    delay_df["source"]
    + "_to_"
    + delay_df["destination"]
)
```

```python
delay_df[
    ["source", "destination", "route"]
].head()
```

```python
delay_df["previous_delay_category"] = pd.cut(
    delay_df["previous_delay"],
    bins=[-1, 5, 15, float("inf")],
    labels=["Low", "Medium", "High"]
)
```

```python
delay_df[
    ["previous_delay", "previous_delay_category"]
].head(10)
```

```python
X_delay = delay_df.drop(
    columns=["delay_minutes"]
)

y_delay = delay_df["delay_minutes"]
```

```python
print("Features:", X_delay.shape)
print("Target:", y_delay.shape)
```

```python
passenger_df = pd.read_csv(
    "../data/processed/passenger_processed.csv"
)

passenger_df.head()
```

```python
passenger_df["is_peak_hour"] = passenger_df["hour"].apply(
    lambda x: 1 if (7 <= x <= 10 or 17 <= x <= 21) else 0
)
```

```python
passenger_df["is_weekend"] = passenger_df[
    "day_of_week"
].apply(
    lambda x: 1 if x >= 5 else 0
)
```

```python
passenger_df["weather_encoded"] = passenger_df[
    "weather"
].map(weather_mapping)
```

```python
X_passenger = passenger_df[
    [
        "hour",
        "day_of_week",
        "holiday",
        "special_event",
        "is_peak_hour",
        "is_weekend",
        "weather_encoded"
    ]
]

y_passenger = passenger_df["passenger_count"]
```

```python
X_train_passenger, X_test_passenger, y_train_passenger, y_test_passenger = train_test_split(
    X_passenger,
    y_passenger,
    test_size=0.2,
    random_state=42
)
```

```python
maintenance_df = pd.read_csv(
    "../data/processed/maintenance_processed.csv"
)

maintenance_df.head()
```

```python
X_maintenance = maintenance_df[
    [
        "engine_temperature",
        "brake_pressure",
        "vibration",
        "wheel_condition",
        "engine_hours",
        "last_service_days",
        "maintenance_count"
    ]
]

y_maintenance = maintenance_df["failure"]
```

```python
X_train_maintenance, X_test_maintenance, y_train_maintenance, y_test_maintenance = train_test_split(
    X_maintenance,
    y_maintenance,
    test_size=0.2,
    random_state=42,
    stratify=y_maintenance
)
```

```python
crowd_df = pd.read_csv(
    "../data/processed/crowd_processed.csv"
)

crowd_df.head()
```

```python
crowd_df["is_peak_hour"] = crowd_df["hour"].apply(
    lambda x: 1 if (7 <= x <= 10 or 17 <= x <= 21) else 0
)
```

```python
crowd_df["is_weekend"] = crowd_df[
    "day_of_week"
].apply(
    lambda x: 1 if x >= 5 else 0
)
```

```python
crowd_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

crowd_df["crowd_level_encoded"] = crowd_df[
    "crowd_level"
].map(crowd_mapping)
```

```python
X_crowd = crowd_df[
    [
        "hour",
        "day_of_week",
        "holiday",
        "train_frequency",
        "passenger_count",
        "entry_count",
        "exit_count",
        "is_peak_hour",
        "is_weekend"
    ]
]

y_crowd = crowd_df["crowd_level_encoded"]
```

```python
X_train_crowd, X_test_crowd, y_train_crowd, y_test_crowd = train_test_split(
    X_crowd,
    y_crowd,
    test_size=0.2,
    random_state=42,
    stratify=y_crowd
)
```
