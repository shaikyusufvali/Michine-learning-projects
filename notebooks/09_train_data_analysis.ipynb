```python
# ============================================================
# RAILPULSE-AI
# 09 - TRAIN DATA ANALYSIS
# ============================================================

import pandas as pd
import os

print("Starting RailPulse-AI Train Data Analysis...")


# ============================================================
# 1. LOAD DATA
# ============================================================

DATA_PATH = "../data/raw/train_data.csv"

print("Current folder:", os.getcwd())
print("Looking for:", os.path.abspath(DATA_PATH))


if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"train_data.csv not found!\n"
        f"Expected path: {os.path.abspath(DATA_PATH)}"
    )


df = pd.read_csv(DATA_PATH)


# ============================================================
# 2. BASIC INFORMATION
# ============================================================

print("\n============================================================")
print("DATASET LOADED")
print("============================================================")

print("Shape:", df.shape)


print("\nColumns:")

for col in df.columns:
    print("-", col)


# ============================================================
# 3. DATA TYPES
# ============================================================

print("\n============================================================")
print("DATA TYPES")
print("============================================================")

print(df.dtypes)


# ============================================================
# 4. FIRST 10 ROWS
# ============================================================

print("\n============================================================")
print("FIRST 10 ROWS")
print("============================================================")

print(df.head(10).to_string(index=False))


# ============================================================
# 5. MISSING VALUES
# ============================================================

print("\n============================================================")
print("MISSING VALUES")
print("============================================================")

print(df.isnull().sum())


# ============================================================
# 6. DUPLICATES
# ============================================================

print("\n============================================================")
print("DUPLICATES")
print("============================================================")

print("Duplicate rows:", df.duplicated().sum())


# ============================================================
# 7. UNIQUE VALUES
# ============================================================

print("\n============================================================")
print("UNIQUE VALUES")
print("============================================================")

for col in df.columns:

    print(
        f"{col}: {df[col].nunique()} unique values"
    )


# ============================================================
# 8. NUMERIC SUMMARY
# ============================================================

print("\n============================================================")
print("NUMERIC SUMMARY")
print("============================================================")

print(
    df.describe().round(2)
)


# ============================================================
# 9. FINAL STATUS
# ============================================================

print("\n============================================================")
print("🎉 09 TRAIN DATA ANALYSIS STARTED SUCCESSFULLY!")
print("============================================================")
```

```python
# ============================================================
# 09 - TARGET & FEATURE ANALYSIS
# ============================================================

print("\n============================================================")
print("TARGET COLUMN ANALYSIS")
print("============================================================")

target = "delay_minutes"

print("Target column:", target)
print("Target data type:", df[target].dtype)

print("\nTarget statistics:")
print(df[target].describe().round(2))

print("\nTarget unique values:")
print(df[target].nunique())

print("\nSample target values:")
print(df[target].head(20).to_list())


print("\n============================================================")
print("CATEGORICAL COLUMNS")
print("============================================================")

categorical_cols = df.select_dtypes(
    include=["object"]
).columns.tolist()

for col in categorical_cols:
    print(f"\n{col}")
    print("Unique values:", df[col].nunique())
    print(df[col].value_counts().head(10))


print("\n============================================================")
print("NUMERICAL COLUMNS")
print("============================================================")

numeric_cols = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

for col in numeric_cols:
    print("-", col)


print("\n============================================================")
print("CORRELATION WITH DELAY MINUTES")
print("============================================================")

numeric_df = df.select_dtypes(
    include=["int64", "float64"]
)

print(
    numeric_df.corr()["delay_minutes"]
    .sort_values(ascending=False)
    .round(3)
)


print("\n============================================================")
print("MISSING VALUES")
print("============================================================")

print(df.isnull().sum())


print("\n============================================================")
print("09 ANALYSIS COMPLETED")
print("============================================================")
```

```python
# ============================================================
# 09 - DATA CLEANING
# ============================================================

print("\n============================================================")
print("DATA CLEANING")
print("============================================================")

# Create a copy
train_df = df.copy()

# ------------------------------------------------------------
# 1. Check missing values
# ------------------------------------------------------------

print("\nMissing values BEFORE cleaning:")
print(train_df.isnull().sum())


# ------------------------------------------------------------
# 2. Remove duplicate rows
# ------------------------------------------------------------

duplicates = train_df.duplicated().sum()

print("\nDuplicate rows:", duplicates)

if duplicates > 0:
    train_df = train_df.drop_duplicates()


# ------------------------------------------------------------
# 3. Handle missing numerical values
# ------------------------------------------------------------

numeric_cols = train_df.select_dtypes(
    include=["number"]
).columns

for col in numeric_cols:
    if train_df[col].isnull().sum() > 0:
        train_df[col] = train_df[col].fillna(
            train_df[col].median()
        )


# ------------------------------------------------------------
# 4. Handle missing categorical values
# ------------------------------------------------------------

categorical_cols = train_df.select_dtypes(
    include=["str"]
).columns

for col in categorical_cols:
    if train_df[col].isnull().sum() > 0:
        train_df[col] = train_df[col].fillna("Unknown")


# ------------------------------------------------------------
# 5. Convert scheduled departure
# ------------------------------------------------------------

train_df["scheduled_departure"] = pd.to_datetime(
    train_df["scheduled_departure"],
    errors="coerce"
)


# ------------------------------------------------------------
# 6. Check target missing values
# ------------------------------------------------------------

train_df = train_df.dropna(
    subset=["delay_minutes"]
)


# ------------------------------------------------------------
# 7. Final check
# ------------------------------------------------------------

print("\nMissing values AFTER cleaning:")
print(train_df.isnull().sum())

print("\nShape BEFORE cleaning:", df.shape)
print("Shape AFTER cleaning :", train_df.shape)

print("\n============================================================")
print("✅ DATA CLEANING COMPLETED")
print("============================================================")
```

```python
# ============================================================
# 09 - FEATURE ENGINEERING
# ============================================================

print("\n============================================================")
print("FEATURE ENGINEERING")
print("============================================================")


# ------------------------------------------------------------
# 1. Extract time-based features
# ------------------------------------------------------------

train_df["departure_hour"] = (
    train_df["scheduled_departure"].dt.hour
)

train_df["departure_minute"] = (
    train_df["scheduled_departure"].dt.minute
)

train_df["departure_day"] = (
    train_df["scheduled_departure"].dt.day
)

train_df["departure_month"] = (
    train_df["scheduled_departure"].dt.month
)

train_df["departure_dayofweek"] = (
    train_df["scheduled_departure"].dt.dayofweek
)


# ------------------------------------------------------------
# 2. Create peak-hour feature
# ------------------------------------------------------------

train_df["is_peak_hour"] = train_df["departure_hour"].apply(
    lambda x: 1 if x in [7, 8, 9, 17, 18, 19, 20] else 0
)


# ------------------------------------------------------------
# 3. Create distance category
# ------------------------------------------------------------

train_df["distance_category"] = pd.cut(
    train_df["distance_km"],
    bins=[0, 100, 300, 600, float("inf")],
    labels=["Short", "Medium", "Long", "Very_Long"]
)


# ------------------------------------------------------------
# 4. Display new features
# ------------------------------------------------------------

print("\nNew features created:")

new_features = [
    "departure_hour",
    "departure_minute",
    "departure_day",
    "departure_month",
    "departure_dayofweek",
    "is_peak_hour",
    "distance_category"
]

for feature in new_features:
    print("-", feature)


# ------------------------------------------------------------
# 5. Show sample
# ------------------------------------------------------------

print("\n============================================================")
print("FEATURE ENGINEERING SAMPLE")
print("============================================================")

display(
    train_df[
        [
            "scheduled_departure",
            "departure_hour",
            "departure_minute",
            "departure_dayofweek",
            "is_peak_hour",
            "distance_km",
            "distance_category",
            "delay_minutes"
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 6. Final dataset information
# ------------------------------------------------------------

print("\n============================================================")
print("FINAL DATASET INFORMATION")
print("============================================================")

print("Shape:", train_df.shape)

print("\nColumns:")
for col in train_df.columns:
    print("-", col)


print("\n============================================================")
print("✅ FEATURE ENGINEERING COMPLETED")
print("============================================================")
```

```python
# ============================================================
# 09 - PREPARE FEATURES AND TARGET
# ============================================================

print("\n============================================================")
print("PREPARING FEATURES AND TARGET")
print("============================================================")


# ------------------------------------------------------------
# 1. Define target
# ------------------------------------------------------------

target = "delay_minutes"


# ------------------------------------------------------------
# 2. Columns to remove
# ------------------------------------------------------------

drop_cols = [
    "train_id",
    "train_name",
    "scheduled_departure",
    target
]


# ------------------------------------------------------------
# 3. Create X and y
# ------------------------------------------------------------

X = train_df.drop(
    columns=drop_cols,
    errors="ignore"
)

y = train_df[target]


# ------------------------------------------------------------
# 4. Convert categorical columns
# ------------------------------------------------------------

categorical_cols = X.select_dtypes(
    include=["object", "category", "str"]
).columns.tolist()

print("\nCategorical columns:")
for col in categorical_cols:
    print("-", col)


# ------------------------------------------------------------
# 5. One-Hot Encoding
# ------------------------------------------------------------

X = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)


# ------------------------------------------------------------
# 6. Make sure all values are numeric
# ------------------------------------------------------------

X = X.apply(
    pd.to_numeric,
    errors="coerce"
)

X = X.fillna(0)


# ------------------------------------------------------------
# 7. Final information
# ------------------------------------------------------------

print("\n============================================================")
print("FEATURES READY")
print("============================================================")

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget:", target)

print("\nFeature columns:")
for col in X.columns:
    print("-", col)


print("\n============================================================")
print("SAMPLE FEATURES")
print("============================================================")

display(X.head())


print("\n============================================================")
print("✅ FEATURES AND TARGET PREPARED")
print("============================================================")
```

```python
# ============================================================
# 09 - TRAIN / TEST SPLIT + MODEL TRAINING
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

print("\n============================================================")
print("TRAIN / TEST SPLIT")
print("============================================================")


# ------------------------------------------------------------
# 1. Split data
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])


# ------------------------------------------------------------
# 2. Create Random Forest model
# ------------------------------------------------------------

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)


# ------------------------------------------------------------
# 3. Train model
# ------------------------------------------------------------

print("\nTraining Random Forest...")

model.fit(
    X_train,
    y_train
)


# ------------------------------------------------------------
# 4. Make predictions
# ------------------------------------------------------------

y_pred = model.predict(X_test)


# ------------------------------------------------------------
# 5. Status
# ------------------------------------------------------------

print("\n============================================================")
print("MODEL TRAINING COMPLETED")
print("============================================================")

print("Model:", type(model).__name__)
print("Trees:", model.n_estimators)

print("\nFirst 10 predictions:")
print(y_pred[:10].round(2))


print("\n============================================================")
print("✅ RANDOM FOREST REGRESSION TRAINED SUCCESSFULLY")
print("============================================================")
```

```python
# ============================================================
# 09 - MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

print("\n============================================================")
print("MODEL EVALUATION")
print("============================================================")


# ------------------------------------------------------------
# 1. Calculate metrics
# ------------------------------------------------------------

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)


# ------------------------------------------------------------
# 2. Display results
# ------------------------------------------------------------

print("\nMean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R² Score:", round(r2, 4))
print("R² Percentage:", round(r2 * 100, 2), "%")


print("\n============================================================")
print("✅ MODEL EVALUATION COMPLETED")
print("============================================================")
```

```python
# ============================================================
# 09 - SAMPLE TRAIN DELAY PREDICTIONS
# ============================================================

print("\n============================================================")
print("SAMPLE TRAIN DELAY PREDICTIONS")
print("============================================================")


# ------------------------------------------------------------
# 1. Create prediction dataframe
# ------------------------------------------------------------

prediction_df = pd.DataFrame({
    "Actual_Delay_Minutes": y_test.values,
    "Predicted_Delay_Minutes": y_pred
})


# ------------------------------------------------------------
# 2. Round values
# ------------------------------------------------------------

prediction_df["Actual_Delay_Minutes"] = (
    prediction_df["Actual_Delay_Minutes"].round(2)
)

prediction_df["Predicted_Delay_Minutes"] = (
    prediction_df["Predicted_Delay_Minutes"].round(2)
)


# ------------------------------------------------------------
# 3. Calculate error
# ------------------------------------------------------------

prediction_df["Error_Minutes"] = (
    prediction_df["Actual_Delay_Minutes"]
    - prediction_df["Predicted_Delay_Minutes"]
)

prediction_df["Absolute_Error_Minutes"] = (
    prediction_df["Error_Minutes"].abs().round(2)
)


# ------------------------------------------------------------
# 4. Display first 20 predictions
# ------------------------------------------------------------

print("\n============================================================")
print("FIRST 20 PREDICTIONS")
print("============================================================")

display(
    prediction_df.head(20)
)


# ------------------------------------------------------------
# 5. Display formatted output
# ------------------------------------------------------------

print("\n============================================================")
print("ACTUAL vs PREDICTED DELAY")
print("============================================================")

print(
    prediction_df.head(20).to_string(index=False)
)


# ------------------------------------------------------------
# 6. Prediction summary
# ------------------------------------------------------------

print("\n============================================================")
print("PREDICTION SUMMARY")
print("============================================================")

print(
    "Average Actual Delay   :",
    round(prediction_df["Actual_Delay_Minutes"].mean(), 2),
    "minutes"
)

print(
    "Average Predicted Delay:",
    round(prediction_df["Predicted_Delay_Minutes"].mean(), 2),
    "minutes"
)

print(
    "Average Absolute Error :",
    round(prediction_df["Absolute_Error_Minutes"].mean(), 2),
    "minutes"
)


print("\n============================================================")
print("✅ SAMPLE PREDICTIONS COMPLETED")
print("============================================================")
```

```python
# ============================================================
# 09 - SAVE MODEL, PREDICTIONS & FEATURE IMPORTANCE
# ============================================================

import os
import joblib

print("\n============================================================")
print("SAVING MODEL AND RESULTS")
print("============================================================")


# ------------------------------------------------------------
# 1. Create directories
# ------------------------------------------------------------

MODEL_DIR = "../models/delay"
PROCESSED_DIR = "../data/processed"

os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(PROCESSED_DIR, exist_ok=True)


# ------------------------------------------------------------
# 2. Save trained model
# ------------------------------------------------------------

MODEL_PATH = os.path.join(
    MODEL_DIR,
    "train_delay_regressor.pkl"
)

joblib.dump(model, MODEL_PATH)


# ------------------------------------------------------------
# 3. Save predictions
# ------------------------------------------------------------

PREDICTION_PATH = os.path.join(
    PROCESSED_DIR,
    "train_delay_predictions.csv"
)

prediction_df.to_csv(
    PREDICTION_PATH,
    index=False
)


# ------------------------------------------------------------
# 4. Feature importance
# ------------------------------------------------------------

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)


# ------------------------------------------------------------
# 5. Save feature importance
# ------------------------------------------------------------

IMPORTANCE_PATH = os.path.join(
    PROCESSED_DIR,
    "train_delay_feature_importance.csv"
)

feature_importance.to_csv(
    IMPORTANCE_PATH,
    index=False
)


# ------------------------------------------------------------
# 6. Display top 10 features
# ------------------------------------------------------------

print("\n============================================================")
print("TOP 10 IMPORTANT FEATURES")
print("============================================================")

display(
    feature_importance.head(10)
)


# ------------------------------------------------------------
# 7. Display saved paths
# ------------------------------------------------------------

print("\n============================================================")
print("MODEL SAVED")
print("============================================================")

print(
    "Model path:",
    os.path.abspath(MODEL_PATH)
)

print(
    "Predictions path:",
    os.path.abspath(PREDICTION_PATH)
)

print(
    "Feature importance path:",
    os.path.abspath(IMPORTANCE_PATH)
)


# ------------------------------------------------------------
# 8. Final status
# ------------------------------------------------------------

print("\n============================================================")
print("🎉 09 - TRAIN DELAY PREDICTION COMPLETED SUCCESSFULLY!")
print("============================================================")
```
